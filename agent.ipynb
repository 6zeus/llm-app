```python
from langchain.agents import AgentType, initialize_agent, load_tools
from langchain_community.llms import VLLM

# 1. llm
## 1.1 replace to model you want use
model = "Qwen/Qwen2.5-1.5B-Instruct"
llm = VLLM(model=model,
           trust_remote_code=True,
           max_new_tokens=520,
           top_k=10,
           top_p=0.95,
           temperature=0.8,
           dtype='half'
           # tensor_parallel_size=... # for distributed inference
           )

# 2. tools
## 2.1 serpapi_api_key for serpapi, apply from https://serpapi.com/
serpapi_api_key = "replace with your own serpapi_api_key"
tools = load_tools(["serpapi", "llm-math"], llm=llm, serpapi_api_key=serpapi_api_key)

# 3. Agent
agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

query = "预测2027年中国人口总数"
result = agent.run(query)
print(f"Result: {result}")
```
